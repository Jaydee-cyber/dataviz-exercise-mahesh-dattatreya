```python
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd

df = pd.read_csv('/Users/maheshjd/data-viz-class-material/data/world_happiness_2023.csv')
# Explore the dataset before you start
print("Region in dataset:")
print(df['Regional indicator'].value_counts())
print("\nScore range:", df['Ladder score'].min(), "–", df['Ladder score'].max())

print("\nBottom 10 countries:")
print(df.nsmallest(10, 'Ladder score')[['Country name','Regional indicator','Ladder score']])
```

```python
region_avg = (df.groupby('Region')['Happiness_Score']
              .mean()
              .reset_index()
              .sort_values('Happiness_Score'))  # sort for horizontal bar

print(region_avg)

fig = px.bar(region_avg,
             x='Happiness_Score',
             y='Region',
             orientation='h',
             title='Average Happiness Score by Region')

fig.show()
```
